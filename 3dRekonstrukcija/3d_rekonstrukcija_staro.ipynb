```python
import numpy as np
import numpy.linalg as la
```

```python
def aff2hom(t):
    return np.append(t, 1)
```

```python
def hom2aff(t):
    if(t[-1] == 0):
        raise ZeroDivisionError("deljenje nulom")
    return t[:-1]/t[-1]
```

```python
def prosek(a,b,c):
    a1 = hom2aff(a)
    b1 = hom2aff(b)
    c1 = hom2aff(c)
    return (a1+b1+c1)/3
```

```python
def osmoTeme(t1, t2, t3, t5, t6, t7, t8):
    p15 = np.cross(t1, t5)
    p26 = np.cross(t2, t6)
    p37 = np.cross(t3, t7)

    Xb1 = np.cross(p15, p26)
    Xb2 = np.cross(p15, p37)
    Xb3 = np.cross(p26, p37)
    Xb = aff2hom(prosek(Xb1, Xb2, Xb3))

    p12 = np.cross(t1, t2)
    p56 = np.cross(t5, t6)
    p78 = np.cross(t7, t8)

    Yb1 = np.cross(p12, p56)
    Yb2 = np.cross(p12, p78)
    Yb3 = np.cross(p56, p78)
    Yb = aff2hom(prosek(Yb1, Yb2, Yb3))

    p23 = np.cross(t2, t3)
    p58 = np.cross(t5, t8)
    p67 = np.cross(t6, t7)

    Zb1 = np.cross(p23, p58)
    Zb2 = np.cross(p23, p67)
    Zb3 = np.cross(p58, p67)
    Zb = aff2hom(prosek(Zb1, Zb2, Zb3))

    p48 = np.cross(Xb, t8)
    p34 = np.cross(Yb, t3)
    p14 = np.cross(Zb, t1)

    t41 = np.cross(p48, p34)
    t42 = np.cross(p48, p14)
    t43 = np.cross(p34, p14)
    t4 = prosek(t41, t42, t43)
    return aff2hom(t4)
```

```python
O3 = np.array([0,0,0,0])
def dveJed(Ah, A1h):
    jed1 = np.concatenate((O3, -A1h[2]*Ah, A1h[1]*Ah)) 
    jed2 = np.concatenate(( A1h[2]*Ah, O3, -A1h[0]*Ah))
    return np.array([jed1, jed2])

def matrica_kamere(pts, pts1):
    A = []
    for i in range(len(pts)):
        dve = dveJed(pts[i], pts1[i])
        A.append(dve[0])
        A.append(dve[1])
    np.array(A)
    U,S, Vh = la.svd(A)   
    matP = Vh[-1].reshape(3,4) 
    return matP/matP[-1,-1] 
```

```python
def centar(T):
    C1 = np.linalg.det(np.delete(T, 0, 1))
    C2 = np.linalg.det(np.delete(T, 1, 1))
    C3 = np.linalg.det(np.delete(T, 2, 1))
    C4 = np.linalg.det(np.delete(T, 3, 1))
    C = np.array([C1, -C2, C3, -C4])*(-1/C4)
    return C
```

```python
def matrica_kalibracije(T):
    t0 = np.delete(T, 3, 1)
    inverz = np.linalg.inv(t0)
    Q, R = np.linalg.qr(inverz)
    if(R[0,0] < 0):
        R = np.matmul(np.diag([-1,1,1]), R)
        q = np.matmul(Q ,np.diag([-1,1,1]))
    if(R[1,1] < 0):
        R = np.matmul(np.diag([1,-1,1]), R)
        q = np.matmul(Q ,np.diag([1,-1,1]))
    if(R[2,2] < 0):
        R = np.matmul(np.diag([1,1,-1]), R)
        q = np.matmul(Q ,np.diag([1,1,-1]))
    
    K = np.linalg.inv(R)
    return K/K[-1,-1]
```

```python
def jed(a, b):
    return np.array([a[0]*b[0], a[1]*b[0], a[2]*b[0], a[0]*b[1], a[1]*b[1], a[2]*b[1], a[0]*b[2], a[1]*b[2], a[2]*b[2]])

def fundamentalna_matrica(l, r):
    A = []
    for i in range(len(l)):
        jna = jed(l[i], r[i])
        A.append(jna)
    np.array(A)
    U,S, Vh = la.svd(A)   
    matP = Vh[-1].reshape(3,3) 
    return matP/matP[-1,-1] 
```

```python
def jednacine(t1, t2, m1, m2):
    j = np.array([
        m1[1]*t1[2] - m1[2]*t1[1],
        -m1[0]*t1[2] + m1[2]*t1[0],
        m2[1]*t2[2] - m2[2]*t2[1],
        -m2[0]*t2[2] + m2[2]*t2[0]
    ])
    return j
def triangulacija(t1, t2, m1, m2):
    jedM = jednacine(t1, t2, m1, m2)
    V = la.svd(jedM)[2]
    return hom2aff(V.T[3])
```
